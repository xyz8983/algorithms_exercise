### step 3 Feature Processing
- step 3.1 convert categorical data into numberical data
- step 3.2 split the dataframe into features and target variable
- step 3.3 split the dataset into training and testing sets


```python
drug_df_encoded = pd.get_dummies(drug_df, columns=["Sex", "BP", "Cholesterol"], drop_first=True)
```

```python
drug_df_encoded.head(5)
```

```python
# try different wayas of encoding categorical variable
```

```python
encoded_df = drug_df_encoded.copy()
```

```python
# label encoding
label_encoder = LabelEncoder()
encoded_df["SexLabel"] = label_encoder.fit_transform(drug_df["Sex"])
```

```python
encoded_df.head(4)
```

```python
# one hot encoding

# one_hot_encoder_F = OneHotEncoder(sparse_output=False)
# one_hot_encoder_T = OneHotEncoder(sparse_output=True)
# BP_encoded_F = one_hot_encoder_F.fit(drug_df["BP"])
# BP_encoded_T = one_hot_encoder_T.fit(drug_df["BP"])
```

```python
transformed_df = drug_df.copy()
le_sex = LabelEncoder()
le_sex.fit(["F", "M"])
transformed_df["Sex"] = le_sex.transform(transformed_df["Sex"])

le_BP = LabelEncoder()
le_BP.fit(['LOW', 'NORMAL', 'HIGH'])
transformed_df["BP"] = le_BP.transform(transformed_df["BP"])

le_Chol = LabelEncoder()
le_Chol.fit(['NORMAL', 'HIGH'])
transformed_df["Cholesterol"] = le_Chol.transform(transformed_df["Cholesterol"])
```

```python
transformed_df.head(4)
```

#### step 3.2 split into features and target columms

```python
X = transformed_df.drop("Drug", axis=1)
Y = transformed_df["Drug"]
```

```python
mapping = dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)))
print(mapping)
```

### step 3.3 split into training and testing set

```python
## Controls the shuffling applied to the data before applying the split. Pass an int for reproducible output across multiple function calls
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=3)
```

### step 4 fit in classification model
- option 1: decision tree

```python
drugTree = DecisionTreeClassifier(criterion="entropy", max_depth = 4)
```

```python
drugTree.fit(X_train,y_train)
```

```python
predTree = drugTree.predict(X_test)
```

```python
print(classification_report(y_test, predTree))
```

#### - option 2: logistic regression

```python
LRclassifier = LogisticRegression(solver='liblinear', max_iter=5000)
LRclassifier.fit(X_train, y_train)
```

```python
y_pred = LRclassifier.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```
