<b> 322 Coin Change </b>

say the proble is F(S), we can make the problem become the composition of the optimal solutions of its sub problems.  
F(S) = F(S-C) + 1  S is the targe value, C is the coin value used.

can sort the coins, and start i from the coins[0] to amount

the meaning of dp, the initial value of dp[0], and how to updated dp[i] are the key to the solution


```python
def coinChange(coins, amount):
    """buttom up, build dp from small amount of value to larger amount of value"""
    # dp[i] is the fewest number of coins making up amount i
    # dp[i] = min(dp[i-coin])+1 for diff coin in coins
    dp = [math.inf] * (amount + 1)
    dp[0] = 0 # for 0 amount, 0 coin is used
    for coin in coins:
        for i in range(coin, amount+1):
            dp[i] = min(dp[i], dp[i-coin]+1) # -> this is the key
    if dp[amount] == math.inf:
        return -1
    else:
        return dp[amount]
```

<b> 518 Coin Change II </b>

also using a dp with length (amount+1), but the meaning of dp[i] is the number of ways to construct amount i in different combination of coins.

how the dp[i] should be updated is smart:
- dp[i] += dp[i-coin]

```python

def change(amount: int, coins: List[int]) -> int:
    """
    dp[i]: the number of ways to make amount i using different combination of coins
    dp[amount] is the result to return
    """

    dp = [0]* (amount+1)
    dp[0] = 1 # exactly 1 way to construct amount 0, which is selecting no coins

    """
    dp[i-coin] there are this amount of way to construct amount i - coin
    so there are this amount (dp[i-coin]) of ways to construct amount i, by adding coin 
    to i-coin.
    dp[i] += dp[i-coin]
    """
    for coin in coins:
        for i in range(coin, amount+1):
            # i starting from coin, because negative amount is invalid
            dp[i] += dp[i-coin]
    return dp[amount]

```

The loop iterates over the candidates list starting from the start index. This ensures that combinations are built without repeating previous val, which prevents duplicates.

```python

def combinationSum(candidates: List[int], target: int) -> List[List[int]]:
    def helper(amount, start, tmp):
        if amount == 0:
            # find a valid combination
            # append the copy version
            res.append(tmp[:]) 
            return
        for i in range(start, len(candidates)):
            if candidates[i] <= amount:
                tmp.append(candidates[i])
                helper(amount - candidates[i], i, tmp)
                tmp.pop() # -> this is the backtracking


    res = []
    helper(target, 0, [])
    return res
```

<b> 1255 Maximum Score Words Formed by Letters </b>

keep track of the counter of left letters that can be used to construct the words. It actually loops through all possible combinations, and keep track of the largest score encoutered.

- also pass in index i as next starting point
- backtracking by undo the previous steps

```python
from collections import Counter

def maxScoreWords(words: List[str], letters: List[str], score: List[int]) -> int:
    if not words or not letters:
        return 0
    letter_count = Counter(letters)

    def helper(letter_count, start_ind):
        """
        return the max scores giving the letter count and starting index 
        """
        max_score = 0
        for i in range(start_ind, len(words)):
            word_score = 0 # score for ith word
            valid_check = True
            for c in words[i]:
                letter_count[c] -= 1
                word_score += score[ord(c) - ord('a')]
                if letter_count[c] < 0:
                    valid_check = False
                    # not stop early, because the reverse part will undo the whole loop of the word[i]

            # backtracking only when word[i] is valid
            if valid_check:
                word_score += helper(letter_count, i+1)
                max_score = max(max_score, word_score)

            # revert to previous status
            for c in words[i]:
                letter_count[c] += 1
                word_score = 0
        return max_score
    return helper(letter_count, 0)

```
