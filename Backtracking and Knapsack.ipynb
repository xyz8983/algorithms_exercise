say the proble is F(S), we can make the problem become the composition of the optimal solutions of its sub problems.  
F(S) = F(S-C) + 1  S is the targe value, C is the coin value used.

can sort the coins, and start i from the coins[0] to amount

the meaning of dp, the initial value of dp[0], and how to updated dp[i] are the key to the solution

time complexity
-   O(N*T) N is the size of coins, T is the amount value, because of the nested for loop

```python
import math
def coinChange(coins, amount):
    """buttom up, build dp from small amount of value to larger amount of value"""
    # dp[i] is the fewest number of coins making up amount i
    # dp[i] = min(dp[i-coin])+1 for diff coin in coins
    dp = [math.inf] * (amount + 1)
    dp[0] = 0 # for 0 amount, 0 coin is used
    for coin in coins:
        for i in range(coin, amount+1):
            # for every coin, update all possible i that this coin can update
            dp[i] = min(dp[i], dp[i-coin]+1) # -> this is the key
    if dp[amount] == math.inf:
        return -1
    else:
        return dp[amount]
```

<b> 518 Coin Change II </b>

also using a dp with length (amount+1), but the meaning of dp[i] is the number of ways to construct amount i in different combination of coins.

how the dp[i] should be updated is smart:
- dp[i] += dp[i-coin]

```python

def change(amount: int, coins: List[int]) -> int:
    """
    dp[i]: the number of ways to make amount i using different combination of coins
    dp[amount] is the result to return
    """

    dp = [0]* (amount+1)
    dp[0] = 1 # exactly 1 way to construct amount 0, which is selecting no coins

    """
    dp[i-coin] there are this amount of way to construct amount i - coin
    so there are this amount (dp[i-coin]) of ways to construct amount i, by adding coin 
    to i-coin.
    dp[i] += dp[i-coin]
    """
    for coin in coins:
        for i in range(coin, amount+1):
            # update every i that is possible of using coin
            # i starting from coin, because negative amount is invalid
            dp[i] += dp[i-coin]
    return dp[amount]

```

for knowing count, the DP way is better because it can reduce duplicated calculations, but to know the exact combinations, the backtracking way is better, even though it is not efficient in time complexity

<b> 39 combination sum </b>

next problem of 518 coin exchange II, this problem wants to know the actual combinations of different ways to construct amount i, not the count.

The loop iterates over the candidates list starting from the start index. This ensures that combinations are built without repeating previous val, which prevents duplicates.

time complexity analysis from leetcode
-   recursive depth is T/min_c, T: target, min_c: smallest candidate val
-   the total number of steps during the backtracking is the number of nodes in the tree. the maximum number of nodes in a N-ary tree of T/min_c height is N ^ (T/min_c) -> each level, at most there are n nodes
-   so the time complexity is O(N^(T/min_c))

space complexity:
-   space overhead for recursion is related to the depth of the recursion step: O(T/min_c)
-   result size: O(2^T) for all valid combination in worse case 
-   each valid combination can take up to T/min_c elements
-   so O(T/min_c + T/min_c*2^T)

time complexity:
-   each element can be used unlimitedly, in worse case the valid combination can be exponential: 2^T
    -   candidate: [1], target = T, for every recursive call, determine if should include 1 or not into the result, this recursion continues until T is reached, so 2^T, the total node number in the recursion tree is O(2^T), even though there is only one solution of: [1]*T, but the process of recursion takes O(2^T)
-   T: target value, N: the number of candidate

-   the solution iterate over N candidate at each step of recursion, 
-   and the number of all nodes of the recursion tree in worse case is 2^T
-   the value is unlimited usage, so every iteration goes over all possible combinations.
-   so total time is O(N * 2^T)

space complexity:
-   recursion depth: max depth = O(T/min_c), min_c is the smallest candidate value
-   result size: O(2^T) for all valid combination in worse case 
-   each valid combination can take up to T/min_c elements
-   in total O(T/min_c * 2^T)

```python

def combinationSum(candidates: List[int], target: int) -> List[List[int]]:
    def helper(amount, start, tmp):
        if amount == 0:
            # find a valid combination
            # append the copy version
            res.append(tmp[:]) 
            return
        for i in range(start, len(candidates)):
            if candidates[i] <= amount:
                tmp.append(candidates[i])
                helper(amount - candidates[i], i, tmp)
                tmp.pop() # -> this is the backtracking


    res = []
    helper(target, 0, [])
    return res
```

<b> 40 combination Sum II </b>

now the combination should be unique, also the char at each position can only be used once, however the char in the candidate can be duplciated. -> return all possible combinations.

the helper function start_ind should keep moving forward, so previously used char is not used again.

```python
class Solution:
    def combinationSum2(self, candidates: List[int], target: int) -> List[List[int]]:
        def helper(res, tmp, remain, start_ind):
            if remain == 0:
                res.append(tmp[:])
                return
            elif remain < 0:
                return
            else:
                for i in range(start_ind, len(candidates)):
                    if i != start_ind and candidates[i]==candidates[i-1]:
                        # avoid duplication
                        continue
                    helper(res, tmp+[candidates[i]], remain - candidates[i], i+1)
                
        candidates.sort()
        res = []
        helper(res, [], target, 0)
        return res

        
```

<b> 1255 Maximum Score Words Formed by Letters </b>

keep track of the counter of left letters that can be used to construct the words. It actually loops through all possible combinations, and keep track of the largest score encoutered.

- also pass in index i as next starting point
- backtracking by undo the previous steps

```python
from collections import Counter

def maxScoreWords(words: List[str], letters: List[str], score: List[int]) -> int:
    if not words or not letters:
        return 0
    letter_count = Counter(letters)

    def helper(letter_count, start_ind):
        """
        return the max scores giving the letter count and starting index 
        """
        max_score = 0
        for i in range(start_ind, len(words)):
            word_score = 0 # score for ith word
            valid_check = True
            for c in words[i]:
                letter_count[c] -= 1
                word_score += score[ord(c) - ord('a')]
                if letter_count[c] < 0:
                    valid_check = False
                    # not stop early, because the reverse part will undo the whole loop of the word[i]

            # backtracking only when word[i] is valid
            if valid_check:
                word_score += helper(letter_count, i+1)
                max_score = max(max_score, word_score)

            # revert to previous status
            for c in words[i]:
                letter_count[c] += 1
                word_score = 0
        return max_score
    return helper(letter_count, 0)

```

<b> 78 subsets </b>

if not allowed to use the itertools.combinations method, this solution utilize the backtracking solutions to list out all the possible substes.

time complexity:
-   each element can be absent or present, so totally 2^N different combinations
-   each element found need O(N) to copy it over to the result
-   so in total O(N*2^N)

space complexity:
-   2^N is the total number of subsets, each subset at most has N value, so worse case it is O(N * 2^N)

```python
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        def helper(res_list, tmp, start, size):
            if len(tmp) == size:
                res_list.append(tmp[:])
            else:
                for i in range(start, len(nums)):
                    tmp.append(nums[i])
                    helper(res_list, tmp, i+1, size)
                    tmp.pop()
        res_list = [[]]
        n = len(nums)
        for i in range(1, n+1):
            helper(res_list, [], 0, i)
        return res_list
```

<b> 90 subsets II </b>

the array contains duplicated number, but the solution should be unique.

can modify the solution of 78 by looping over size then adding size check in the helper, which is more intuitive.

```python
class Solution:
    def subsetsWithDup(self, nums: List[int]) -> List[List[int]]:
        def helper(res_list, tmp, s):
            res_list.append(tmp[:])
            for i in range(len(s)):
                # this steps avoid the duplicated subset
                if i > 0 and s[i] == s[i-1]:
                    continue
                helper(res_list, tmp+[s[i]], s[i+1:])
        res_list = []
        nums.sort() # this is important, otherwise the s[i]==s[i-1] check is meaningless
        helper(res_list, [], nums)
        return res_list
```
