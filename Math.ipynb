<b> 670. Maximum Swap</b>

search from backwards, find the largest value found so far, track its index, then keep moving backward, to find the furtherest index that has value smaller than this found largest value, then you find a pair of index for swapping. If during this process, a larger value is found, then updating the largest index, but the pair of index is only updated when a smaller value is found in the front.

```python

def maximumSwap(num: int) -> int:
    """
    from right to left, find the largest value, mark as max_ind
    then from max_ind to left, find the left most position that is smaller than this 
    largest value. 
    in this way after the swap, the largest value can be placed as left as 
    possible hence generating the largest value

    the max_ind may keep updating, without having a valid left_ind, 
    like when left_ind is smaller than the max, so here whenever the left_ind is found, snapshot the max_ind to right_ind
    """
    str_num = list(str(num))
    max_ind = len(str_num) - 1
    left_ind = right_ind = 0
    for i in range(len(str_num)-1, -1, -1):
        # find the right most largest value's index 
        if str_num[i] > str_num[max_ind]:
            max_ind = i
        # find a possible swapping pair
        elif str_num[i] < str_num[max_ind]:
            left_ind = i
            right_ind = max_ind
    # if no swapping pair found, left_ind and right_ind are all 0, swapping for the same value does not change the value
    str_num[left_ind], str_num[right_ind] = str_num[right_ind], str_num[left_ind]
    return int("".join(str_num))
```

```python
# max_ind at the end is 0, but the pair of swapping pair is (1, 2)
maximumSwap(736)
```

<b> 31 Next Permutation </b>

lexicographic order.

```python

def nextPermutation(nums: List[int]) -> None:
    """
    Do not return anything, modify nums in-place instead.
    
    exlanation: 
    find the first element from right to the left that breaks the acending trend, this ith element is what we want to change
    then find the first element from the right to the left that is greater than the ith element located above, there will always be such one from the right because all values right to ith are in acending order until ith. that is jth element

    switch the ith and jth element. Now all values to the ith are in ascending order from the right to the left (remember how we locate i and j), that is the largest order for the subset from i+1: to the end. we need to reverse it to get its smallest order. The order upgrade is already made by replacing ith with jth element.
    """
    #if the elements are increasing from the right, they are currently at their largest possible permutation
    i = len(nums)-2
    while i>=0 and nums[i]>=nums[i+1]:
        i -= 1
    if i>=0:
        j = len(nums)-1
        while j >=0 and nums[j]<=nums[i]:
            j-=1  # find the first element from the right (lower priority in lexicographical order) that is greater than nums[i], using greater value to switch with i so the order is increasing after switching
        
        nums[i], nums[j] = nums[j], nums[i]
    # the position to the right of i is in increasing order from right to left, not in the smallest order version, so need to reverse it.
    nums[i+1: ] = nums[i+1: ][::-1]
        
```

<b> 66 Plus one </b>

add one to a number represented in list. smart to just check the 9

```python

def plusOne(digits):
    """
    :type digits: List[int]
    :rtype: List[int]
    """
    
    len_s = len(digits)
    for i in range(len_s-1, -1, -1):
        if digits[i] < 9:
            digits[i]+=1
            return digits
        else:
            digits[i] = 0
    # if not return yet, reaching this step meaning the res should be 100....
    digits.append(0)
    digits[0]=1
    return digits
```
