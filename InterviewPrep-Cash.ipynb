### <b> OOD, family tree </b>
- Represent this family tree
- traverse through this family tree, given an input string as a person, find their parents, grandparents,and siblings.
- one Possible way: several methods to find the relatives


```python
from typing import List
class Person:
    def __init__(self, *, name:str, children: List =[], parents: List = []):
        self.name = name
        self.children = set(children)
        self.parents = set(parents)
        
    def add_children(self, children: List[Person]):
        self.children = self.children | set(children)
        
    def find_parents(self):
        return [p.name for p in self.parents]
    
    def find_grantparents(self):
        # return 4 ppl
        grand_parents = []
        for parent in self.parents:
            grand_parents += parent.parents
        return [p.name for p in grand_parents]
    
    def find_siblings(self):
        # if share the same parents / or if share one of the parent
        siblings = set()
        for parent in self.parents:
            siblings = siblings | set(parent.children)
        return [p.name for p in siblings if p!=self]
        
    
```

```python
a = Person(name="a")
b = Person(name="b")
c = Person(name="c")

d = Person(name="d", parents=[a, b])
a.add_children([d])
b.add_children([d])
e = Person(name="e", parents=[a, b])
a.add_children([e])
b.add_children([e])

l = Person(name="l", parents=[b,c])
b.add_children([l])
c.add_children([l])
               

f = Person(name="f")
g = Person(name="g")
h = Person(name="h")

i = Person(name="i", parents=[f, g])
j = Person(name="j", parents=[f, g])
f.add_children([i, j])
g.add_children([i, j])

k = Person(name="k", parents=[i, d])
i.add_children([k])
d.add_children([k])
```

### <b> OOD, implement an event listener </b>  
should support add new listener, release listener, delete listener   
- function 1: register listener for a specific event
- function 2: Emit events, which should run corresponding event listeners registered.
- example: Creates an instance of an event emitter. A = newEventEmitter()
- example: Registers an event specific listener. A.addListener(event, listener)
- example: Runs all registered event listeners for the specific event.: a.emit(event)
- example: Removes the given listener from the given event: a.removeListener(event, listener)

```python
# suppose event: string, listener: string id
from collections import defaultdict
class Event():
    def __init__(self, name: str):
        self.name = name
        
    
class EventEmitter():
    # connection between event and listner, decouple the action from the object class
    def __init__(self):
        self.eventToListener = defaultdict(set) # one event can have multiple listners
        
    def registerEventWithListener(self, event: Event, listener: str):
        self.eventToListener[event].add(listener)
        
    def emitEvent(self, event: Event):
        # returns all registered event listener
        return self.eventToListener[event]
    
    def removeListener(self, event: Event,listener:str):
        self.eventToListener[event].remove(listener)
        
    
        
        
```

```python
a = Event("a")
b = Event("b")
eventEmitter = EventEmitter()
eventEmitter.registerEventWithListener(a, "l1")
eventEmitter.registerEventWithListener(a, "l2")
eventEmitter.registerEventWithListener(b, "l1")

print(eventEmitter.emitEvent(a))
eventEmitter.removeListener(a, "l1")
print(eventEmitter.emitEvent(a))
print(eventEmitter.emitEvent(b))
```

### OOD string wrapper api
- step 1: 模拟一个输入输出都是字符串的API，这个API有50%的概率返回正确结果，25%的概率返回transient error，25%的概率返回fatal error。实现一个这样的buggy API。


```python
import random
import requests
def getString(input_string: str)->str: 
    random_num = random.randint(1, 100)
    if random_num <=50:
        return input_string
    elif 50< random_num <= 75:
        raise requests.exceptions.Timeout # as transiente error
        # or requests.exceptions.HTTPError
    else:
        raise ValueError
```

- step 2: 写一个wrapper来wrap这个API，正确结果直接返回，fatal error直接结束，transient error要重试N次，直接重试不需要backoff设计

```python
import logging
logger = logging.getLogger(__name__)

def triggerGetString(input_string: str, max_retry: int):
    retry_count = 0
    while retry_count <= max_retry:
        try:
            return getString(input_string)
        except requests.exceptions.Timeout:
            # retry n times
            logger.info(f"retry {retry_count} time for timeout")
            retry_count += 1
        except ValueError as e:
            logger.error(e)
            break
        
        
        
```

```python
for i in range(100):
    print(triggerGetString("heyThere", 3))
```

- step 3: 在原先的buggy API里增加一个error的状态：不能立刻重试的状态比如throttled error。正确、transient、throttle、fatal每个都有1/4的概率。
- throttle error: http response error cod 429, try too many requests

```python
import random
import requests
def getString(input_string: str)->str: 
    random_num = random.randint(1, 100)
    if random_num <=25:
        return input_string
    elif 25 <random_num <=50:
        raise IndexError # as the throttling error
    elif 50< random_num <= 75:
        raise requests.exceptions.Timeout # as transiente error
        # or requests.exceptions.HTTPError
    else:
        raise ValueError
```

- step 4: 在wrapper里为throttled error加上linear backoff。tricky的地方在于如果是连续的throttled error，要backoff时间增加（第一次是1s，第二次是2s……）；如果中间收到了transient error，则下次遇到throttled error的backoff的时间回到1s。第二步规定的总共重试的时间不变。


```python
import logging
import time
logger = logging.getLogger(__name__)

def triggerGetString(input_string: str, max_retry: int):
    retry_count = 0
    sleep_backoff_time = 0
    exception_name = ""
    
    while retry_count <= max_retry:
        try:
            time.sleep(sleep_backoff_time)
            return getString(input_string)
        except IndexError as e:
            # as a fake throttle error
            if exception_name == "throttle_error":
                sleep_backoff_time = 2 * sleep_backoff_time
            else:
                exception_name = "throttle_error"
                sleep_backoff_time = 1
                
            logger.info(f"retry {retry_count} time for throttle error")
            retry_count += 1
            
        except requests.exceptions.Timeout:
            # retry n times
            exception_name = "transient_error"
            logger.info(f"retry {retry_count} time for timeout")
            retry_count += 1
        except ValueError as e:
            exception_name = "fatal_error"
            logger.error(e)
            break   
        
```

### series of Q
- give a word: square, and other similar words like Square, Squars, find one that only has one char typo
    - if upper case / lower case difference is considered as typo?
    - is one char more or less considered as typo?


```python
def check_one_char_diff(source, targets):
    n = len(source)
    for word in targets:
        m = len(word)
        diff_count = 0
        l = min(n, m)
        for i in range(l):
            if source[i]!=word[i]:
                diff_count += 1
                if diff_count > 1:
                    break
        
        diff_count += max(n-l, m-l)
        if diff_count == 1:
            return word
    return ""
```

```python
check_one_char_diff("square", ["Square", "Squars"])
```

- given a word: square, and other similar words like swuare, because w and q are closed to each other in keyboard, this type of typo is called: close typo. Return a word that is the close typo of "square"
  - q: how many close typo is allowed?
  - q: only left and right char in keyboard is considered as close typo? no up and down char?
  - q: length of two words should be the same?


```python
# define a dictionary that key is each char in keyboard, value is the set of adjacent char
# that can be considered as close typo
close_typo_dict = {"w": {"q"}, "q": {"w"}}
def check_one_char_diff(source, targets):
    n = len(source)
    for word in targets:
        m = len(word)
        close_typo_count = 0
        if m!=n:
            continue
        for i in range(n):
            if source[i]!=word[i] and word[i] in close_typo_dict[source[i]]:
                close_typo_count += 1
                if close_typo_count > 1:
                    break
            elif source[i]!=word[i]:
                break
        if close_typo_count == 1:
            return word # or add to a res list

    return ""
```

```python
check_one_char_diff("square", ["swuare", "Square"])
```
