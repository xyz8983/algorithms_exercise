<b> 1101 the eariliest time when everyone becomes friends </b>

using union find.  
find method: find the final parent  
union: if two does not have same parent, then connect them  
the group size is essential, initial value is N meaning there are separate N group, whenever two ppl are connected by union method, the group size reduced. When group size is 1 that means everyone is in same group, then time found (here should sort the original list by timestamp)

There are n people in a social group labeled from 0 to n - 1. You are given an array logs where logs[i] = [timestampi, xi, yi] indicates that xi and yi will be friends at the time timestampi.

Friendship is symmetric. That means if a is friends with b, then b is friends with a. Also, person a is acquainted with a person b if a is friends with b, or a is a friend of someone acquainted with b.

Return the earliest time for which every person became acquainted with every other person. If there is no such earliest time, return -1.

```python
class Solution(object):
    def earliestAcq(self, logs, n):
        """
        :type logs: List[List[int]]
        :type n: int
        :rtype: int
        """
        # sort the logs by timestamps
        # union element who has same parent
        # whe there is only one parent, it is end

        self.group_size = n # intially every friend is separated, so n group
        self.groups = {i: i for i in range(n)}

        def find(x):
            if self.groups[x]!=x:
                # x:x means finding the parent, to the end
                # if not means x has parent, keep moving up and update x's parent with grand parent
                self.groups[x] = find(self.groups[x])
            return self.groups[x]
        def union(x, y):
            x, y = find(x), find(y)
            if (x != y):
                # separate group size should reduce
                self.group_size -= 1
                # connect x and y
                self.groups[x] = y
        # sort by timestamps
        for t, x, y in sorted(logs):
            union(x, y)
            if self.group_size == 1:
                # only one group left, and since logs is already sorted, the time should be ealiest
                return t
        return -1
```

```python
logs = [[20190101,0,1],[20190104,3,4],[20190107,2,3],[20190211,1,5],[20190224,2,4],[20190301,0,3],[20190312,1,2],[20190322,4,5]]
n = 6
s = Solution()
s.earliestAcq(logs, n)
```

<b> 547 number of province </b>

similar to above. only call the union function when city i and j are connected

There are n cities. Some of them are connected, while some are not. If city a is connected directly with city b, and city b is connected directly with city c, then city a is connected indirectly with city c.

A province is a group of directly or indirectly connected cities and no other cities outside of the group.

You are given an n x n matrix isConnected where isConnected[i][j] = 1 if the ith city and the jth city are directly connected, and isConnected[i][j] = 0 otherwise.

Return the total number of provinces.



```python
class Solution:
    def findCircleNum(self, isConnected):
        """
        :type isConnected: List[List[int]]
        :rtype: int
        """
        n = len(isConnected)
        self.groups = {i: i for i in range(n)}
        # worse case every city is not connected to each other so n province
        self.province_num = n

        def find(x):
            # if groups[x]==x -> that is the initial status, no other connection
            if self.groups[x]!=x:
                # find its root connection recursively
                self.groups[x] = find(self.groups[x])
            return self.groups[x]

        def union(x, y):
            x, y = find(x), find(y)
            if x != y:
                # means x and y do not share any connection from previous record
                # now add them to one province by connecting them
                self.groups[x] = y
                # since x and y city now belongs to one provice, the inital province num should reduce
                self.province_num -= 1
        # only need to loop over the matrix in right above the diagnol of the matrix
        for i in range(n-1):
            for j in range(i+1,n):
                # only when city i and j are connected, then to union them to see if it belongs to a province already
                if isConnected[i][j] == 1:
                    union(i, j)
        return self.province_num
```

```python
isConnected = [[1,1,0],[1,1,0],[0,0,1]]
s = Solution()
s.findCircleNum(isConnected)
```

<b> 737 sentence similarities II </b>

extra nodes that needs to be parsed first.

construct the set of vocabularies, initialize the connection to each word themselves. first going over the similarites pair to construct the connection. then compare sentence 1 and sentence 2 word by word, if the two words are similar, they should find the same connection which is constructed previously

Problems:


We can represent a sentence as an array of words, for example, the sentence "I am happy with leetcode" can be represented as arr = ["I","am",happy","with","leetcode"].

Given two sentences sentence1 and sentence2 each represented as a string array and given an array of string pairs similarPairs where similarPairs[i] = [xi, yi] indicates that the two words xi and yi are similar.

Return true if sentence1 and sentence2 are similar, or false if they are not similar.

Two sentences are similar if:

They have the same length (i.e., the same number of words)
sentence1[i] and sentence2[i] are similar.
Notice that a word is always similar to itself, also notice that the similarity relation is transitive. For example, if the words a and b are similar, and the words b and c are similar, then a and c are similar.

```python
class Solution(object):
    def areSentencesSimilarTwo(self, sentence1, sentence2, similarPairs):
        """
        :type sentence1: List[str]
        :type sentence2: List[str]
        :type similarPairs: List[List[str]]
        :rtype: bool
        """
        word_set_1 = set()
        for x, y in similarPairs:
            word_set_1.add(x)
            word_set_1.add(y)
        all_words = set(sentence1).union(set(sentence2)).union(word_set_1)
        self.groups = {w: w for w in all_words}

        def find(x):
            if self.groups[x]!=x:
                self.groups[x] = find(self.groups[x])
            return self.groups[x]

        def union(x, y):
            x, y = find(x), find(y)
            if x!=y:
                self.groups[x]=y
        
        if len(sentence1)!=len(sentence2):
            return False
        else:
            for x, y in similarPairs:
                # update the groups connection 
                union(x, y)

            n = len(sentence1)
            for i in range(n):
                if find(sentence1[i])!=find(sentence2[i]):
                    return False
            return True
```

```python
sentence1 = ["great","acting","skills"]
sentence2 = ["fine","drama","talent"]
similarPairs = [["great","good"],["fine","good"],["drama","acting"],["skills","talent"]]
s = Solution()
s.areSentencesSimilarTwo(sentence1, sentence2, similarPairs)
```

<b> sentence similarity </b>

here is a simplified version of problems that the similarity is not transitive. meaning a->b , b -> c, DOES NOT mean a -> c

We can represent a sentence as an array of words, for example, the sentence "I am happy with leetcode" can be represented as arr = ["I","am",happy","with","leetcode"].

Given two sentences sentence1 and sentence2 each represented as a string array and given an array of string pairs similarPairs where similarPairs[i] = [xi, yi] indicates that the two words xi and yi are similar.

Return true if sentence1 and sentence2 are similar, or false if they are not similar.

Two sentences are similar if:

They have the same length (i.e., the same number of words)
sentence1[i] and sentence2[i] are similar.
Notice that a word is always similar to itself, also notice that the similarity relation is not transitive. For example, if the words a and b are similar, and the words b and c are similar, a and c are not necessarily similar.

```python
class Solution(object):
    def areSentencesSimilar(self, sentence1, sentence2, similarPairs):
        """
        :type sentence1: List[str]
        :type sentence2: List[str]
        :type similarPairs: List[List[str]]
        :rtype: bool
        """
        similarity_pair_set = {(x, y) for x, y in similarPairs}
        if len(sentence1)!= len(sentence2):
            return False
        else:
            n = len(sentence1)
            for i in range(n):
                w1 = sentence1[i]
                w2 = sentence2[i]
                # note than 2 words are definitly similar if they are the same
                if (w1!=w2) and ((w1, w2) not in similarity_pair_set) and ((w2, w1) not in similarity_pair_set):
                    return False
            return True
```

<b> 2102 detonate the maximum bombs </b>

more a graph question, and actually not solved using union find patter.  
first construct a graph with key bomb point to a list of bomb that it can detonate.  
then loop over each node to keep track of how many node it can detonate by going over the graph in dfs manner.  then compare the the count to get max number


```python
from collections import defaultdict
class Solution(object):
    def maximumDetonation(self, bombs):
        """
        :type bombs: List[List[int]]
        :rtype: int
        """

        self.graph = defaultdict(list)
        n = len(bombs)
        # construct a graph of bomb: [list of bomb in its range]
        for i in range(n):
            for j in range(n):
                if i == j: continue
                # j is in i's bomb range
                if (bombs[i][0]-bombs[j][0])**2 + (bombs[i][1]-bombs[j][1])**2 <= bombs[i][2]**2:
                    self.graph[i].append(j)

        def dfs(bomb_ind, visited):
            # node: the bomb
            # a set of bomb that can be triggered by this node bomb
            # finally altered the visited set
            for related_bomb_ind in self.graph[bomb_ind]:
                if related_bomb_ind not in visited:
                    visited.add(related_bomb_ind)
                    dfs(related_bomb_ind, visited)

        res = 0
        for i in range(n):
            # initiated a set for each child, must contain itself as the first one to explode
            visited = set([i])
            dfs(i, visited)
            res = max(res, len(visited))
        return res

            
```

```python
bombs=[[1,2,3],[2,3,1],[3,4,2],[4,5,3],[5,6,4]]
s = Solution()
s.maximumDetonation(bombs)
```
