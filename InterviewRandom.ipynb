### print out org structure

given a list of list of name, first name of the nested list is the manager, rest of names in the nested list are the direct reports of the manager.

for example:  
input: [[a,b,c],  
		[b,d],  
		[d,e]]  

<b> Q1: print out the org chart, like:   </b>
a  
....b  
........d  
............e  
....c  

It is a dfs search problem, first construct a graph with:  
{ manager: [direct reports]}, and when looping over the graph, keep track of the level as int. if it is in same level, the number of "." in front of the name should be the same.

```python
from typing import List
from collections import defaultdict
def print_org_chart(names: List[List[str]], head:str):
    graph = defaultdict(list)
    for team in names:
        graph[team[0]] = team[1:]
    
    def dfs(node, level):
        org_line = "...." * level + node
        print(org_line)
        
        # not a manager
        if node not in graph:
            return
        for direct_report in graph[node]:
            dfs(direct_report, level+1)
    dfs(head, 0)
```

```python
print_org_chart([["a","b", "c"], ["b", "d"], ["d", "e"]], "a")
```

### print the connection

			input: ['B', ['C', 'D'], ['E', ['F', ['G', 'H']]]]
			output：print:  
			# B  
			# B->C  
			# B->D  
			# B->C->E  
			# B->D->E  
			# B->C->E->F  
			# B->D->E->F  
			# B->C->E->F->G  
			# B->D->E->F->G  
			# B->C->E->F->H  
			# B->D->E->F->H  


```python
def dfs(input_list, res: List[str]):
    output = []
    for item in input_list:
        
        if type(item) == str:
            
            if not res:
                output.append(item)
                print(item)
            else:
                for s in res:
                    new_s = s + "->" + item
                    output.append(new_s)
                    print(new_s)
            
        elif type(item) == list:
            output = dfs(item, output)
    return output
```

```python
res = dfs(['B', ['C', 'D'], ['E', ['F', ['G', 'H']]]], [])
```
