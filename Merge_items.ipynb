```python
def mergeAlternately(word1: str, word2: str) -> str:
    s = min(len(word1), len(word2))
    res = []
    for i in range(s):
        res.append(word1[i])
        res.append(word2[i])
    res_str = ''.join(res)
    return res_str+word1[s:]+word2[s:]
```

```python
from collections import deque
class ZigzagIterator:
    """
    using a queue to save the list, pop a list one by one, after using it, append it to the end for next round of usage
    """
    def __init__(self, v1: List[int], v2: List[int]):
        self.queue = deque()
        if v1: self.queue.append((len(v1), iter(v1)))
        if v2: self.queue.append((len(v2), iter(v2)))

    def next(self) -> int:
        count, item_list = self.queue.popleft()
        res = next(item_list)
        count -= 1
        if count > 0:
            self.queue.append((count, item_list))
        return res


    def hasNext(self) -> bool:
        return len(self.queue) > 0
        

# Your ZigzagIterator object will be instantiated and called as such:
# i, v = ZigzagIterator(v1, v2), []
# while i.hasNext(): v.append(i.next())
```


<b> 88 merge sorted array </b>  
O(M+N) in time.  remember to check the if nums2 is not finished placing after the while loop

```python

def merge(nums1: List[int], m: int, nums2: List[int], n: int) -> None:
    """
    Do not return anything, modify nums1 in-place instead.
    """
    target = len(nums1)-1
    i = m-1
    j = n-1
    while i>=0 and j>=0:
        if nums1[i] > nums2[j]:
            nums1[target] = nums1[i]
            i-=1
        else:
            nums1[target] = nums2[j]
            j-=1
        target -=1
    if j>=0:
        nums1[:j+1]=nums2[:j+1]
    return nums1
```

<b> 21 merge two sorted list </b>

create a dummy head, and its copy, then use this copy to chain the two list while comparing them, it is like 3 poiters, one (copy head) indicates the current location for insertion, l1 for position of list 1, l2 for position of list 2

time complexity:
- O(N+M) N and M are the size of the two list

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def mergeTwoLists(self, l1: ListNode, l2: ListNode) -> ListNode:
        start = ListNode()
        node = start   # important, node is point to start in initial step
        while l1 and l2:
            if l1.val<=l2.val:
                node.next = l1
                l1 = l1.next
            else:
                node.next = l2
                l2 = l2.next 
            node = node.next
        if l1:
            node.next = l1
        if l2:
            node.next = l2
        return start.next
```

<b> 23 merge K sorted list </b>

now instead of just merging 2 sorted list, merging K sorted list.

since comparing k list with each other is not optimal, here we can put all items from all list into a priority (node.val, id(node), node), the second value is used to break the tie when two node has the same value.

then we can chain the item one by one by order.

here we construct the queue by inserting the first element of each of the list, so each time the priority queue's length is small, roughly the size of lists. 

time complexity:
- initial creation of the priority takes O(MlogM) 
- all item loop once: O(N) N is the total amount of nodes
- O(logM) for each pop operation from priority queue. 
- in total O(NlogM) + O(MlogM) = O(NlogM)

space complexity:
- O(M) the priority queue size is roughly the size of the lists, M element, each element from one of the lists

```python
import heapq
def mergeKLists(lists: List[ListNode]) -> ListNode:
    q = []
    dummy = ListNode()
    curr = dummy
    for i in lists:
        # add id(node) incase the node has same val and break the queue, since listNode itself does not have comparison method
        if i: heapq.heappush(q, (i.val, id(i), i))
    while q:
        _, _, node = heapq.heappop(q)
        curr.next = node
        curr = node
        if node.next:
            heapq.heappush(q, (node.next.val, id(node.next), node.next))
    return dummy.next     
```

<b> 148 sort list </b>

sort a single linked list

it reuses the solution of problem 21, and use divide and concur to break the original list into half, then do the merge.

double check the base case, should include head.next too otherwise it may duplicate the result

time complexity:
- cost of splitting the list into half totally cost O(NlogN) N is the size of the list for all recursive levels.
- for each merge action of two list, it visit all nodes from the two lists only once, so total process of all merges across all recursion level is O(N), and there are logN level, so O(NlogN)
- splitting in half + merging is 2(NlogN) so O(NlogN)

(one split takes N, one merge takes N, and there are logN recursive call, so O(NlogN))

space complexity:
- each recursive call takes constant space, and there are logN level, so O(logN) in total 

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def sortList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        """
        divide and concur
        sort sub list, then merge the sublist
        """
        # base, empty node or a single node
        if not head or not head.next:
            return head
        # cut the list to haf
        prev = None
        slow = head
        fast = head
        while fast and fast.next:
            prev = slow
            slow = slow.next
            fast = fast.next.next
        prev.next = None

        # then sort each half
        l1 = self.sortList(head) # starting head of first half
        l2 = self.sortList(slow)

        # merge l1 and l2
        return self.merge(l1, l2)
    
    def merge(self, l1: Optional[ListNode], l2: Optional[ListNode]):
        new_head = ListNode()
        prev = new_head
        while l1 and l2:
            if l1.val < l2.val:
                prev.next = l1
                l1 = l1.next
            else:
                prev.next = l2
                l2 = l2.next
            prev = prev.next
        
        if l1:
            prev.next = l1
        if l2:
            prev.next = l2
        return new_head.next

        
```

<b> 147 insertion sort list </b>

now given a linked list, using insertion sort, different than the divide and concur way used in above problem 148.

first find the node that is not in the right position,  
then find the right position to insert this node.

time complexity:
- O(N) for first while loop that only loop once from head to tail
- O(N) roughly N for nested while loop to find the right position, it is O(i) for each while loop. 
- O(N(N-1)/2) = O(N*N) in total

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def insertionSortList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        """
        first find the element that need to adjust position
        then starting from beginning to find the right position
        """
        if not head:
            return
        dummy = ListNode(0, head)
        
        # pointer head only loop once and end until reaching tail
        while head and head.next:
            if head.val > head.next.val:
                # node found
                node = head.next
                # now find the right location
                prev = dummy
                while prev.next.val < node.val:
                    prev = prev.next
                # now prev is the first node with val > node.val
                # node should be inserted before prev node
                # head -> node, prev -> another_node
                # should be prev -> node -> another_node, head -> node.next
                head.next = node.next
                tmp = prev.next
                prev.next = node
                node.next = tmp
            else:
                # moving forward
                head = head.next
        return dummy.next
            
```

<b> 708 insert into a sorted circular linked list </b>

now the linked list is a circular one.

the given head can be any node in the linked list, not necessarily the real head. given the insertVal, insert this value to maintain the sorted list, return value should be the originally given node.  if it is a newly created list then return the new node

the list supposed to be in ascending order, so the next node value should be > current node, otherwise it is the end of the linked list, then we can compare to see if the new node is greater than tail or smaller than head.



```python
"""
# Definition for a Node.
"""
class Node:
    def __init__(self, val=None, next=None):
        self.val = val
        self.next = next

class Solution:
    def insert(self, head: 'Optional[Node]', insertVal: int) -> 'Node':
        """
        linked list so it is a loop
        1) edge case, empty list
        2) prev <= new <= next
        3) new < min (insert at tail)
        4) new > max (insert at tail)
            prev >= new <= next
            prev <= new >= next
        5) all nodes are the same
        """
        node = Node(insertVal)
        if not head:
            node.next = node
            return node
        start = head
        prev = start
        def insert_helper(node, new_node):
            tmp = node.next
            node.next = new_node
            new_node.next = tmp

        while True:
            if start.val <= insertVal <= start.next.val or\
            start.val > start.next.val and insertVal >= start.val or \
            start.val > start.next.val and insertVal <= start.next.val:
                insert_helper(start, node)
                break
            
            start = start.next
            if start == head:
                # loop once already
                insert_helper(start, node)
                break
        return head

        
```

<b> 2. add two numbers </b>

also in linked list. using a adds on to keep check of the value that needs to be add to next digit

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def addTwoNumbers(self, l1: ListNode, l2: ListNode) -> ListNode:
        addup = 0
        start = ListNode()
        node = start  
        while l1 or l2:
            x = l1.val if l1 else 0
            y = l2.val if l2 else 0
            val = x+y+addup
            if val >= 10:
                addup = 1
                val = val%10
            else:
                addup = 0
            node.next = ListNode(val)
            node = node.next  # move to next None
            l1=l1.next if l1 else None
            l2=l2.next if l2 else None
        if addup>0:
            node.next = ListNode(addup)
        return start.next
```

<b> 445 add two numbers II </b>

now the linked list order differently, the first number is the most significant digit (the largest digit), which is different from problem 2 where the first node in the linked list is the smallest digit.

still need to start from the smallest digit, so here using extra space to save the nodes from the two linked list, then pop the node from 2 list which is in the correct order for processing.

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def addTwoNumbers(self, l1: ListNode, l2: ListNode) -> ListNode:
        l1_vals = []
        l2_vals = []
        while l1:
            l1_vals.append(l1.val)
            l1 = l1.next
        while l2:
            l2_vals.append(l2.val)
            l2 = l2.next 
            
        carry_on = 0
        tail_node = None
        while l1_vals or l2_vals or carry_on:
            val1 = l1_vals.pop() if l1_vals else 0
            val2 = l2_vals.pop() if l2_vals else 0
            val = (val1+val2+carry_on)%10
            carry_on = (val1+val2+carry_on)//10
   
            node = ListNode(val)
            node.next = tail_node
            tail_node = node

        return tail_node
```

<b> 43 Multiple string </b>

the adds on way can also be applied to multiple too. the smart part is the save the product at a list at i+j+1 position while looping over the two strings.

```python

def multiply(num1: str, num2: str) -> str:
    l1 = len(num1)
    l2 = len(num2)
    # the final result's length wont be greater than l1+l2
    product = [0]*(l1+l2)  
    for i in range(l1-1, -1, -1):
        for j in range(l2-1, -1, -1):
            x1 = ord(num1[i]) - ord("0")
            x2 = ord(num2[j]) - ord("0")
            # start with l1+l2-1, end of the product array
            # += because multipliaction nature, like 123*456, 5,3 and 2,6, should save to same place for sum later
            product[i+j+1] += x1*x2
    carry = 0
    for i in range(len(product)-1, -1, -1):
        # adds up and update the product array, right most position is the smallest position, like in real multiplication
        temp = (product[i]+carry) % 10
        carry = (product[i]+carry) //10
        product[i] = str(temp)

    result = ("".join(product)).lstrip("0")
    return result if result else "0"
    
```

<b> 67 Add binary, 415 add strings </b>

same as add in binary format or numbers in string

```python

def addBinary( a: str, b: str) -> str:
    n1 = len(a)-1
    n2 = len(b)-1
    result = []
    carry_on = 0
    while n1>=0 or n2>=0:
        x1 = int(a[n1]) if n1>=0 else 0
        x2 = int(b[n2]) if n2>=0 else 0
        value = (x1+x2+carry_on)%2
        carry_on = (x1+x2+carry_on)//2
        result.append(value)
        n1-=1
        n2-=1
    if carry_on:
        result.append(carry_on)
    
    return "".join(str(x) for x in result[::-1])
```
