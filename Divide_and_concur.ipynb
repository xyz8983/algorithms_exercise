<b> 494 Target Sum</b>

quite classic problem. how many different way to use array reaching the target sum, it can be split into sub problem and using a DP to avoid duplicated calculation

the (i, curr_sum) acts like a checkpoint.

```python
class Solution:
    def findTargetSumWays(self, nums: List[int], target: int) -> int:
        # split the question into sub problemes, when one item is taken, the 
        # rest index and leftoever target becomes the new nums and target

        l = len(nums)
        curr_sum = 0
        # key: (i, curr_sum) 
        # val: how many way to reach target sum starting at status (i, curr_sum)
        mem = {} 

        # return how many ways that can reach target from the status: index i with cumulative_sum: curr_sum
        def dp_helper(i, curr_sum, mem):
            # base case
            if (i, curr_sum) in mem:
                return mem[(i, curr_sum)]
            # when reaching the end of array and aggregated sum equals to target
            if i == l and curr_sum == target:
                return 1
            # when reaching the end of array but curr_sum does not match target
            if i==l:
                return 0
            
            #recursive case, have two ways, either positive or negative
            positive = dp_helper(i+1, curr_sum + nums[i], mem)
            negative = dp_helper(i+1, curr_sum - nums[i], mem)
            count = positive + negative
            mem[(i, curr_sum)] = count
            return positive + negative

        return dp_helper(0, curr_sum, mem)
```
