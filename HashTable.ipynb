<b> 1711 count good meals </b>

if directly looping over the list of numbers and check if two numbers are power of two, the power calculation seems time consuming and easily get a exceed time limit error.

instead keeping a hashmap with k as the item from the array, val as the count. because the problems limit the range of each item from the array to be within [2 ** 0, 2 ** 20], we can loop over k from 0 to 21 (2 items add up can be > 2 ** 20) and check if the leftover value already visited or not.

solution with Exceed Time Limit error

```python
import math
from typing import List

def countPairs(deliciousness: List[int]) -> int:
    res = 0 
    n = len(deliciousness)
    visited_valid = set()
    visited_invalid = set()
    if n <=1:
        return 0
    for i in range(n):
        for j in range(i+1, n):
            if (i, j) in visited_valid or (j, i) in visited_valid:
                res += 1
            elif (i, j) in visited_invalid or (j, i) in visited_invalid:
                continue
            else:
                try: 
                    power = math.log2(deliciousness[i] + deliciousness[j])
                    if int(power) == power:
                        res += 1
                        visited_valid.add((i, j))
                    else:
                        visited_invalid.add((i, j))
                except ValueError:
                    visited_invalid.add((i, j))
                    continue
    return res % (10**9 + 7)
```

solution with power check in a hashtable

```python
from collections import defaultdict

def countPairs(deliciousness: List[int]) -> int:
    res = 0
    freq = defaultdict(int)
    for x in deliciousness:
        for k in range(22):
            # 2**0 <= x <= 2**20, so largest k visited can be 21 because two x adds up
            res += freq[2**k - x]
        freq[x] += 1
    return res % (10**9+7)
```

<b> 167 two sum II </b>

using dictionary to save the value to index mapping. pay attention to the return order. if two loop, one for constructing the map, another for finding the item, then [ind+1, map[item]+1] if only one loop that contruct the map and find the target at the same time, then [map[item]+1, ind+1] because the item found in the map is what previously constructed and small value is added to the map first

```python

def twoSum(numbers, target):
    """
    :type numbers: List[int]
    :type target: int
    :rtype: List[int]
    """
    hash_map = {}
    for ind, n in enumerate(numbers):
        val = target - n
        if val in hash_map:
            return [hash_map[val]+1, ind+1]
        else:
            hash_map[n] = ind
```

```python
twoSum([2,7,11,13], 9)
```

<b> 1002 find common characters </b>

smart way of using Counter, &= operaters for Counters, and Counter.elements() function

&= operation supports Sets and Counters. For counters, it keeps only the elements that are present in both Counters, with their minimum counts.

Counter.elements() function returns an iterator over elements repeating each as many times as its count. 

```python
from collections import Counter

def commonChars(words: List[str]) -> List[str]:
    res = Counter(words[0])
    for w in words:
        # To keep only the elements that are present in both Counters, with their minimum counts.
        res &= Counter(w) 
    return list(res.elements())
```

```python
commonChars(["bella","label","roller"])
```
